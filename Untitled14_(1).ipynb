```python
# CELL 5 — Process NIH → Unified Dataset

def process_NIH():
    mapping = {"Parasitized":"Ring", "Uninfected":"Uninfected_RBC"}
    all_files, all_labels = [], []

    for folder, cls in mapping.items():
        fpath = os.path.join(NIH_DIR, folder)
        if not os.path.exists(fpath):
            print(f"⚠️ Not found: {fpath}"); continue
        for f in os.listdir(fpath):
            if f.lower().endswith(('.png','.jpg','.jpeg')):
                all_files.append(os.path.join(fpath, f))
                all_labels.append(cls)

    print(f"Total NIH images: {len(all_files):,}")

    X_tv, X_test, y_tv, y_test = train_test_split(
        all_files, all_labels,
        test_size=0.15, stratify=all_labels, random_state=42)
    X_train, X_val, y_train, y_val = train_test_split(
        X_tv, y_tv,
        test_size=0.176, stratify=y_tv, random_state=42)

    def save_split(paths, labels, split):
        n = 0
        for src, cls in tqdm(zip(paths, labels),
                              total=len(paths),
                              desc=f"Saving {split}"):
            img = cv2.imread(src)
            if img is None: continue
            img  = cv2.resize(img, IMG_SIZE)
            name = "NIH_" + os.path.basename(src).rsplit('.',1)[0]+'.png'
            cv2.imwrite(f"{UNIFIED_DIR}/{split}/{cls}/{name}", img)
            n += 1
        return n

    tr = save_split(X_train, y_train, "train")
    vl = save_split(X_val,   y_val,   "val")
    ts = save_split(X_test,  y_test,  "test_NIH")

    print(f"\n✅ NIH Processed!")
    print(f"   Train    : {tr:,}")
    print(f"   Val      : {vl:,}")
    print(f"   Test_NIH : {ts:,}")

process_NIH()
```

```python
# CELL 7 — Process MP-IDB (DatasetNinja format)
import json

MPIDB_IMG_DIR = "/content/MP-IDB/ds/img"
MPIDB_ANN_DIR = "/content/MP-IDB/ds/ann"

# DatasetNinja label mapping
DN_MAP = {
    "ring":         "Ring",
    "rings":        "Ring",
    "trophozoite":  "Trophozoite",
    "trophozoites": "Trophozoite",
    "schizont":     "Schizont",
    "schizonts":    "Schizont",
    "gametocyte":   "Gametocyte",
    "gametocytes":  "Gametocyte",
    "leukocyte":    "Leukocyte",
    "wbc":          "Leukocyte",
    "rbc":          "Uninfected_RBC",
    "uninfected":   "Uninfected_RBC",
    "healthy":      "Uninfected_RBC",
    "difficult":    "Ring",
    "difficult_rbc":"Ring",
}

def process_MPIDB_ninja():
    ann_files = [f for f in os.listdir(MPIDB_ANN_DIR)
                 if f.endswith('.json')]
    print(f"Total annotation files: {len(ann_files)}")

    processed = 0
    skipped   = 0
    cls_count = {c:0 for c in ALL_CLASSES}
    gidx      = 0

    for ann_file in tqdm(ann_files, desc="MP-IDB"):
        ann_path = os.path.join(MPIDB_ANN_DIR, ann_file)
        img_name = ann_file.replace('.json', '')
        img_path = os.path.join(MPIDB_IMG_DIR, img_name)

        # Try different extensions
        if not os.path.exists(img_path):
            for ext in ['.png', '.jpg', '.jpeg']:
                if os.path.exists(img_path + ext):
                    img_path = img_path + ext
                    break
        if not os.path.exists(img_path):
            skipped += 1; continue

        img = cv2.imread(img_path)
        if img is None: skipped += 1; continue
        h, w = img.shape[:2]

        try:
            with open(ann_path, 'r') as f:
                ann_data = json.load(f)
        except: skipped += 1; continue

        objects = ann_data.get('objects', [])
        if not objects:
            # Whole image = Uninfected
            cell  = cv2.resize(img, IMG_SIZE)
            split = "train" if gidx%20<14 else \
                    ("val"  if gidx%20<17 else "test_MPIDB")
            cv2.imwrite(
                f"{UNIFIED_DIR}/{split}/Uninfected_RBC/"
                f"MP_{img_name}_{gidx}.png", cell)
            cls_count["Uninfected_RBC"] += 1
            processed += 1; gidx += 1
            continue

        for i, obj in enumerate(objects):
            # Get label
            cls_raw = obj.get('classTitle','').lower().strip()
            label   = DN_MAP.get(cls_raw)
            if not label: continue

            # Get bounding box
            exterior = (obj.get('points',{})
                          .get('exterior', []))
            if len(exterior) < 2: continue

            xs = [p[0] for p in exterior]
            ys = [p[1] for p in exterior]
            x1 = max(0,   int(min(xs))-15)
            y1 = max(0,   int(min(ys))-15)
            x2 = min(w-1, int(max(xs))+15)
            y2 = min(h-1, int(max(ys))+15)
            if (x2-x1)<20 or (y2-y1)<20: continue

            cell  = cv2.resize(img[y1:y2, x1:x2], IMG_SIZE)
            split = "train" if gidx%20<14 else \
                    ("val"  if gidx%20<17 else "test_MPIDB")
            cv2.imwrite(
                f"{UNIFIED_DIR}/{split}/{label}/"
                f"MP_{img_name}_{i}.png", cell)
            cls_count[label] += 1
            processed += 1; gidx += 1

    print(f"\n✅ MP-IDB Processed!")
    print(f"   Total cells : {processed:,}")
    print(f"   Skipped     : {skipped}")
    print(f"\n   Class Distribution:")
    for cls, n in cls_count.items():
        flag = " ← RARE ⚠️" if cls in RARE_CLASSES and n < 200 else ""
        print(f"   {cls:<20}: {n:>5}{flag}")

process_MPIDB_ninja()
```

```python
# CELL 7c — Fixed MP-IDB Processing (Bitmap Mask Format)
import json, base64, zlib
import numpy as np

# Species → Stage mapping (medical knowledge based)
SPECIES_MAP = {
    "malariae":        "Ring",
    "falciparum":      "Ring",
    "vivax":           "Trophozoite",
    "ovale":           "Trophozoite",
    "knowlesi":        "Ring",
    "ring":            "Ring",
    "trophozoite":     "Trophozoite",
    "schizont":        "Schizont",
    "gametocyte":      "Gametocyte",
    "leukocyte":       "Leukocyte",
    "wbc":             "Leukocyte",
    "rbc":             "Uninfected_RBC",
    "uninfected":      "Uninfected_RBC",
    "healthy":         "Uninfected_RBC",
    "red blood cell":  "Uninfected_RBC",
    "infected":        "Ring",
}

def decode_bitmap(bitmap_data, origin, img_shape):
    """Decode DatasetNinja bitmap mask → bounding box"""
    try:
        z    = base64.b64decode(bitmap_data)
        data = zlib.decompress(z)
        n    = int(np.frombuffer(data[:4], dtype='>u4')[0])
        m    = int(np.frombuffer(data[4:8], dtype='>u4')[0])
        mask = np.unpackbits(
                   np.frombuffer(data[8:], dtype=np.uint8)
               )[:n*m].reshape(n, m)

        rows = np.any(mask, axis=1)
        cols = np.any(mask, axis=0)
        if not rows.any(): return None

        rmin, rmax = np.where(rows)[0][[0,-1]]
        cmin, cmax = np.where(cols)[0][[0,-1]]

        x1 = max(0, origin[0] + cmin - 15)
        y1 = max(0, origin[1] + rmin - 15)
        x2 = min(img_shape[1]-1, origin[0] + cmax + 15)
        y2 = min(img_shape[0]-1, origin[1] + rmax + 15)
        return x1, y1, x2, y2
    except:
        return None

def process_MPIDB_v2():
    ann_dir = "/content/MP-IDB/ds/ann"
    img_dir = "/content/MP-IDB/ds/img"

    ann_files = [f for f in os.listdir(ann_dir) if f.endswith('.json')]
    print(f"Total annotation files: {len(ann_files)}")

    processed = 0
    skipped   = 0
    cls_count = {c:0 for c in ALL_CLASSES}
    gidx      = 0

    for ann_file in tqdm(ann_files, desc="MP-IDB v2"):
        ann_path = os.path.join(ann_dir, ann_file)
        img_name = ann_file.replace('.json','')
        img_path = os.path.join(img_dir, img_name)

        if not os.path.exists(img_path):
            for ext in ['.jpg','.png','.jpeg']:
                if os.path.exists(img_path+ext):
                    img_path = img_path+ext; break
        if not os.path.exists(img_path):
            skipped+=1; continue

        img = cv2.imread(img_path)
        if img is None: skipped+=1; continue

        with open(ann_path) as f:
            data = json.load(f)

        objects = data.get('objects', [])

        if not objects:
            # Whole image → Uninfected
            cell  = cv2.resize(img, IMG_SIZE)
            split = "train" if gidx%20<14 else \
                    ("val"  if gidx%20<17 else "test_MPIDB")
            cv2.imwrite(
                f"{UNIFIED_DIR}/{split}/Uninfected_RBC/"
                f"MPv2_{img_name}_{gidx}.png", cell)
            cls_count["Uninfected_RBC"]+=1
            processed+=1; gidx+=1; continue

        for i, obj in enumerate(objects):
            cls_raw = obj.get('classTitle','').lower().strip()
            label   = SPECIES_MAP.get(cls_raw)
            if not label: continue

            geo  = obj.get('geometryType','')
            bbox = None

            if geo == 'bitmap':
                bmp    = obj.get('bitmap',{})
                origin = bmp.get('origin',[0,0])
                bbox   = decode_bitmap(
                             bmp.get('data',''),
                             origin, img.shape)

            elif geo == 'rectangle':
                pts = obj.get('points',{}).get('exterior',[])
                if len(pts)>=2:
                    xs=[p[0] for p in pts]
                    ys=[p[1] for p in pts]
                    bbox=(max(0,int(min(xs))-15),
                          max(0,int(min(ys))-15),
                          min(img.shape[1]-1,int(max(xs))+15),
                          min(img.shape[0]-1,int(max(ys))+15))

            elif geo in ['polygon','polyline']:
                pts = obj.get('points',{}).get('exterior',[])
                if len(pts)>=2:
                    xs=[p[0] for p in pts]
                    ys=[p[1] for p in pts]
                    bbox=(max(0,int(min(xs))-15),
                          max(0,int(min(ys))-15),
                          min(img.shape[1]-1,int(max(xs))+15),
                          min(img.shape[0]-1,int(max(ys))+15))

            if bbox is None: continue
            x1,y1,x2,y2 = bbox
            if (x2-x1)<10 or (y2-y1)<10: continue

            cell  = cv2.resize(img[y1:y2,x1:x2], IMG_SIZE)
            split = "train" if gidx%20<14 else \
                    ("val"  if gidx%20<17 else "test_MPIDB")
            cv2.imwrite(
                f"{UNIFIED_DIR}/{split}/{label}/"
                f"MPv2_{img_name}_{i}.png", cell)
            cls_count[label]+=1; processed+=1; gidx+=1

    print(f"\n✅ MP-IDB v2 Done!")
    print(f"   Total cells : {processed:,}")
    print(f"   Skipped     : {skipped}")
    print(f"\n   Class Distribution:")
    for cls,n in cls_count.items():
        flag=" ← RARE ⚠️" if cls in RARE_CLASSES and n<200 else ""
        print(f"   {cls:<20}: {n:>5}{flag}")

process_MPIDB_v2()
```

```python
# CELL 7e — FINAL MP-IDB Fix (PNG bitmap masks)
import json, base64
import numpy as np
from PIL import Image
import io

SPECIES_MAP = {
    "malariae":"Ring",      "falciparum":"Ring",
    "vivax":"Trophozoite",  "ovale":"Trophozoite",
    "knowlesi":"Ring",      "ring":"Ring",
    "trophozoite":"Trophozoite", "schizont":"Schizont",
    "gametocyte":"Gametocyte",   "leukocyte":"Leukocyte",
    "wbc":"Leukocyte",           "rbc":"Uninfected_RBC",
    "uninfected":"Uninfected_RBC","healthy":"Uninfected_RBC",
    "infected":"Ring",
}

def decode_png_mask(b64_data, origin, img_shape):
    """DatasetNinja bitmap = base64 PNG mask"""
    try:
        png_bytes = base64.b64decode(b64_data)
        mask = np.array(Image.open(io.BytesIO(png_bytes)))

        # Get bounding box from mask
        if mask.ndim == 3:
            mask = mask[:,:,0]
        rows = np.any(mask > 0, axis=1)
        cols = np.any(mask > 0, axis=0)
        if not rows.any(): return None

        rmin, rmax = np.where(rows)[0][[0,-1]]
        cmin, cmax = np.where(cols)[0][[0,-1]]

        PAD = 20
        x1 = max(0, origin[0] + cmin - PAD)
        y1 = max(0, origin[1] + rmin - PAD)
        x2 = min(img_shape[1]-1, origin[0] + cmax + PAD)
        y2 = min(img_shape[0]-1, origin[1] + rmax + PAD)
        return int(x1), int(y1), int(x2), int(y2)
    except Exception as e:
        return None

def process_MPIDB_final():
    ann_dir = "/content/MP-IDB/ds/ann"
    img_dir = "/content/MP-IDB/ds/img"
    ann_files = [f for f in os.listdir(ann_dir)
                 if f.endswith('.json')]
    print(f"Total files: {len(ann_files)}")

    processed = 0; skipped = 0
    cls_count = {c:0 for c in ALL_CLASSES}
    gidx = 0

    for ann_file in tqdm(ann_files, desc="MP-IDB Final"):
        ann_path = os.path.join(ann_dir, ann_file)
        img_name = ann_file.replace('.json','')
        img_path = os.path.join(img_dir, img_name)

        if not os.path.exists(img_path):
            for ext in ['.jpg','.png','.jpeg']:
                if os.path.exists(img_path+ext):
                    img_path=img_path+ext; break
        if not os.path.exists(img_path):
            skipped+=1; continue

        img = cv2.imread(img_path)
        if img is None: skipped+=1; continue

        with open(ann_path) as f:
            data = json.load(f)

        objects = data.get('objects',[])

        if not objects:
            cell = cv2.resize(img, IMG_SIZE)
            split="train" if gidx%20<14 else \
                  ("val"  if gidx%20<17 else "test_MPIDB")
            cv2.imwrite(
                f"{UNIFIED_DIR}/{split}/Uninfected_RBC/"
                f"MPf_{img_name}_{gidx}.png", cell)
            cls_count["Uninfected_RBC"]+=1
            processed+=1; gidx+=1; continue

        for i, obj in enumerate(objects):
            cls_raw = obj.get('classTitle','').lower().strip()
            label   = SPECIES_MAP.get(cls_raw)
            if not label: continue

            geo  = obj.get('geometryType','')
            bbox = None

            if geo == 'bitmap':
                bmp    = obj.get('bitmap',{})
                origin = bmp.get('origin',[0,0])
                bbox   = decode_png_mask(
                             bmp.get('data',''),
                             origin, img.shape)

            elif geo in ['rectangle','polygon','polyline']:
                pts = obj.get('points',{}).get('exterior',[])
                if len(pts)>=2:
                    xs=[p[0] for p in pts]
                    ys=[p[1] for p in pts]
                    bbox=(max(0,int(min(xs))-15),
                          max(0,int(min(ys))-15),
                          min(img.shape[1]-1,int(max(xs))+15),
                          min(img.shape[0]-1,int(max(ys))+15))

            if bbox is None: continue
            x1,y1,x2,y2 = bbox
            if (x2-x1)<10 or (y2-y1)<10: continue

            cell = cv2.resize(img[y1:y2,x1:x2], IMG_SIZE)
            split="train" if gidx%20<14 else \
                  ("val"  if gidx%20<17 else "test_MPIDB")
            cv2.imwrite(
                f"{UNIFIED_DIR}/{split}/{label}/"
                f"MPf_{img_name}_{i}.png", cell)
            cls_count[label]+=1; processed+=1; gidx+=1

    print(f"\n✅ MP-IDB Final Done!")
    print(f"   Total cells : {processed:,}")
    print(f"   Skipped     : {skipped}")
    print(f"\n   Class Distribution:")
    for cls,n in cls_count.items():
        flag=" ← RARE ⚠️" if cls in RARE_CLASSES and n<200 else ""
        print(f"   {cls:<20}: {n:>5}{flag}")

process_MPIDB_final()
```

```python
# CELL 15 — Oversample rare classes using augmentation
import cv2, os, numpy as np
from tqdm import tqdm

TARGET = 2000   # target per class

AUG_CLASSES = {
    "Ring"       : 1044,
    "Schizont"   : 380,
    "Gametocyte" : 312,
    "Leukocyte"  : 206,
}

def augment_image(img):
    """Random flip, rotate, brightness, zoom"""
    # Flip
    if np.random.rand() > 0.5:
        img = cv2.flip(img, 1)
    if np.random.rand() > 0.5:
        img = cv2.flip(img, 0)
    # Rotate
    angle = np.random.uniform(-30, 30)
    M = cv2.getRotationMatrix2D((64,64), angle, 1.0)
    img = cv2.warpAffine(img, M, (128,128))
    # Brightness
    factor = np.random.uniform(0.7, 1.3)
    img = np.clip(img * factor, 0, 255).astype(np.uint8)
    return img

for cls, current in AUG_CLASSES.items():
    src_dir  = f"{UNIFIED_DIR}/train/{cls}"
    files    = os.listdir(src_dir)
    needed   = TARGET - current
    if needed <= 0: continue

    print(f"Augmenting {cls}: {current} → {TARGET} (+{needed})")
    for i in tqdm(range(needed), desc=cls):
        src  = os.path.join(src_dir, np.random.choice(files))
        img  = cv2.imread(src)
        aug  = augment_image(img)
        cv2.imwrite(f"{src_dir}/aug_{i}.png", aug)

print("\n✅ Augmentation Done!")

# Final count
print("\n📊 Final Class Distribution (train):")
for cls in ALL_CLASSES:
    n = len(os.listdir(f"{UNIFIED_DIR}/train/{cls}"))
    print(f"   {cls:<20}: {n:>6}")
```

```python
# CELL 18 — All 5 Model Definitions
from torchvision.models import efficientnet_b0, vit_b_16
from torchvision.models import ViT_B_16_Weights

# 1. CNN Only
class CNNBaseline(nn.Module):
    def __init__(self):
        super().__init__()
        base = models.mobilenet_v2(pretrained=True)
        base.classifier = nn.Identity()
        self.backbone = base
        self.head = nn.Sequential(nn.Dropout(0.3), nn.Linear(1280, NUM_CLASSES))
    def forward(self, x): return self.head(self.backbone(x))

# 2. CNN + LSTM
class CNNLSTMModel(nn.Module):
    def __init__(self):
        super().__init__()
        base = models.mobilenet_v2(pretrained=True)
        base.classifier = nn.Identity()
        self.cnn  = base
        self.lstm = nn.LSTM(1280, 256, num_layers=2, batch_first=True, dropout=0.3)
        self.head = nn.Sequential(nn.Dropout(0.3), nn.Linear(256, NUM_CLASSES))
    def forward(self, x):
        feat = self.cnn(x).unsqueeze(1)
        out, _ = self.lstm(feat)
        return self.head(out[:,-1,:])

# 3. ResNet50
class ResNetBaseline(nn.Module):
    def __init__(self):
        super().__init__()
        base = models.resnet50(pretrained=True)
        base.fc = nn.Linear(2048, NUM_CLASSES)
        self.model = base
    def forward(self, x): return self.model(x)

# 4. EfficientNet-B0
class EfficientNetBaseline(nn.Module):
    def __init__(self):
        super().__init__()
        base = efficientnet_b0(pretrained=True)
        base.classifier[1] = nn.Linear(1280, NUM_CLASSES)
        self.model = base
    def forward(self, x): return self.model(x)

# 5. CNN + ViT Hybrid (MAIN NOVEL MODEL)
class CNNViTHybrid(nn.Module):
    def __init__(self):
        super().__init__()
        cnn_base = models.mobilenet_v2(pretrained=True)
        self.cnn = nn.Sequential(*list(cnn_base.features))
        self.cnn_pool    = nn.AdaptiveAvgPool2d((1,1))
        self.cnn_project = nn.Linear(1280, 512)
        vit_base = vit_b_16(weights=ViT_B_16_Weights.DEFAULT)
        vit_base.heads = nn.Identity()
        self.vit = vit_base
        self.vit_project = nn.Linear(768, 512)
        self.attn_gate = nn.Sequential(nn.Linear(1024,2), nn.Softmax(dim=-1))
        self.fusion = nn.Sequential(
            nn.Linear(1024,512), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(512, 256), nn.GELU(), nn.Dropout(0.2),
            nn.Linear(256, NUM_CLASSES)
        )
    def forward(self, x):
        x_vit = nn.functional.interpolate(x,(224,224),mode='bilinear',align_corners=False)
        cnn_feat = self.cnn_pool(self.cnn(x)).flatten(1)
        cnn_feat = self.cnn_project(cnn_feat)
        vit_feat = self.vit_project(self.vit(x_vit))
        combined = torch.cat([cnn_feat, vit_feat], dim=1)
        return self.fusion(combined)

print("✅ All 5 models defined!")
```

```python
# CELL 19 — Training Function
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {DEVICE}")

def train_model(model, name, epochs=15):
    model = model.to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    criterion = nn.CrossEntropyLoss()
    best_acc  = 0.0

    for epoch in range(epochs):
        model.train()
        total_loss = 0
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(imgs), labels)
            loss.backward(); optimizer.step()
            total_loss += loss.item()

        model.eval()
        correct = total = 0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
                preds = model(imgs).argmax(dim=1)
                correct += (preds==labels).sum().item()
                total   += labels.size(0)

        val_acc = correct/total*100
        if val_acc > best_acc:
            best_acc = val_acc
            torch.save(model.state_dict(), f"/content/{name}_best.pth")

        scheduler.step()
        print(f"[{name}] Ep {epoch+1:02d}/{epochs} | Loss: {total_loss/len(train_loader):.4f} | Acc: {val_acc:.2f}%")

    print(f"✅ {name} Best: {best_acc:.2f}%\n")
    return model, best_acc

print("✅ Train function ready!")
```

```python
# CELL 20 — Train all 5 models
results = {}

for model, name in [
    (CNNBaseline(),          "CNN_Only"),
    (CNNLSTMModel(),         "CNN_LSTM"),
    (ResNetBaseline(),       "ResNet50"),
    (EfficientNetBaseline(), "EfficientNet_B0"),
    (CNNViTHybrid(),         "CNN_ViT_Hybrid"),
]:
    print(f"\n{'='*50}\n  {name}\n{'='*50}")
    m, best = train_model(model, name, epochs=15)
    results[name] = best

print("\n" + "="*45)
print("📊 FINAL BENCHMARK RESULTS")
print("="*45)
best_model = max(results, key=results.get)
for name, acc in results.items():
    trophy = " 🏆" if name == best_model else ""
    print(f"  {name:<22}: {acc:.2f}%{trophy}")
```
